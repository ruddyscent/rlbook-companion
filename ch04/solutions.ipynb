## Exercise 4.1: 격자 세계의 행동 가치

예제 4.1의 4×4 격자 세계에서 각 전이의 보상은 $-1$, 종료까지는 $\gamma = 1$이며, 균등 무작위 정책 $\pi$ 아래에서 그림 4.1의 마지막 행이 $v_\pi$를 보여 준다. 행동가치는

$$q_\pi(s, a) = \sum_{s', r} p(s', r \mid s, a)\bigl[r + \gamma\, v_\pi(s')\bigr]$$

로 계산한다. 격자 세계에서 전이는 결정적이므로 합은 한 항으로 줄어든다.

**$q_\pi(11, \texttt{down})$.** 상태 11에서 아래로 한 칸 이동하면 종단 상태에 도달한다. 종단 상태의 가치는 $0$이므로

$$q_\pi(11, \texttt{down}) = -1 + 1 \cdot 0 = -1.$$

**$q_\pi(7, \texttt{down}).$** 상태 7에서 아래로 이동하면 상태 11에 도달한다. 그림 4.1의 마지막 행에 따르면 $v_\pi(11) = -14$이므로

$$q_\pi(7, \texttt{down}) = -1 + 1 \cdot (-14) = -15.$$

## Exercise 4.2: 새 상태 15 추가

예제 4.1의 격자에 새 상태 15를 상태 13 바로 아래에 두고, 그 행동 left, up, right, down이 각각 12, 13, 14, 15로 이어지도록 한다. $\gamma = 1$이고 모든 전이의 보상은 $-1$이며, 균등 무작위 정책에서 각 행동의 확률은 $1/4$이다.

**원래 전이를 그대로 둘 때.** 기존 상태 1–14의 값은 변하지 않는다. 그림 4.1로부터 $v_\pi(12) = -22$, $v_\pi(13) = -20$, $v_\pi(14) = -14$를 가져와 벨만 방정식에 대입하면

$$v_\pi(15) = \tfrac{1}{4}\bigl[(-1 - 22) + (-1 - 20) + (-1 - 14) + (-1 + v_\pi(15))\bigr] = -15 + \tfrac{1}{4}\,v_\pi(15).$$

정리하면 $\tfrac{3}{4}\,v_\pi(15) = -15$이므로 $v_\pi(15) = -20$이다.

**상태 13의 down도 15로 향하도록 바꿀 때.** 상태 13의 한 행동만 바뀌었으므로

$$v_\pi(13) = \tfrac{1}{4}\bigl[(-1 + v_\pi(9)) + (-1 + v_\pi(12)) + (-1 + v_\pi(14)) + (-1 + v_\pi(15))\bigr]$$

이 된다. 만약 $v_\pi(13) = -20$, $v_\pi(15) = -20$이라면 $v_\pi(15)$의 식과 $v_\pi(13)$의 식이 모두 성립하는지 확인해 보자. 우선 위 식에 $v_\pi(9) = -20$, $v_\pi(12) = -22$, $v_\pi(14) = -14$, $v_\pi(15) = -20$을 대입하면

$$v_\pi(13) = \tfrac{1}{4}\bigl[-21 - 23 - 15 - 21\bigr] = -20.$$

$v_\pi(15)$ 쪽도 마찬가지로 $-20$이 다시 나오므로, $(v_\pi(13), v_\pi(15)) = (-20, -20)$은 두 식의 합동 해이다. 격자의 다른 상태들이 모두 변동 없이 같은 값을 유지하므로 이 해가 곧 정답이다. 즉 두 경우 모두 $v_\pi(15) = -20$이며, 이번 변화로 상태 13의 가치도 바뀌지 않는다.

이 결과는 사실 대칭성으로 한눈에 이해할 수 있다. 상태 15는 상태 13의 거울상이고, 두 상태의 값이 같다면 상태 13의 down 행동이 자기 자신으로 머물든 상태 15로 가든 받는 기댓값은 똑같기 때문이다.

## Exercise 4.3: 행동가치에 대한 식 (4.3)–(4.5)

정의에 따라 $q_\pi(s, a) = \mathbb{E}_\pi[G_t \mid S_t = s, A_t = a]$이고, 다음 시점부터는 정책 $\pi$를 따르므로 $G_{t+1}$의 기댓값은 다음 행동 $A_{t+1}$의 가치 $q_\pi(S_{t+1}, A_{t+1})$이 된다. 식 (4.3)에 대응하는 형태는 다음과 같다.

$$q_\pi(s, a) = \mathbb{E}_\pi\bigl[R_{t+1} + \gamma\, q_\pi(S_{t+1}, A_{t+1}) \,\bigm|\, S_t = s,\ A_t = a\bigr]. \tag{4.3$'$}$$

기댓값을 환경 동역학과 정책으로 풀어 쓰면 식 (4.4)에 해당한다.

$$q_\pi(s, a) = \sum_{s', r} p(s', r \mid s, a)\Bigl[r + \gamma \sum_{a'} \pi(a' \mid s')\, q_\pi(s', a')\Bigr]. \tag{4.4$'$}$$

이를 갱신 규칙으로 옮기면 식 (4.5)에 대응하는 행동가치 반복 정책 평가가 나온다.

$$q_{k+1}(s, a) \doteq \sum_{s', r} p(s', r \mid s, a)\Bigl[r + \gamma \sum_{a'} \pi(a' \mid s')\, q_k(s', a')\Bigr]. \tag{4.5$'$}$$

$v_\pi$가 존재하는 같은 조건 아래에서 임의의 $q_0$로 출발한 수열 $\{q_k\}$는 $q_\pi$로 수렴한다.

## Exercise 4.4: 정책 반복의 무한 순환 방지

본문 의사코드의 결함은 정책 개선 단계에서 $\arg\max_a$가 동률을 임의로 푸는 데서 비롯한다. 두 행동 $a_1, a_2$가 $q_\pi$에서 같은 값을 가지면 한 반복에서는 $a_1$, 다음 반복에서는 $a_2$가 선택될 수 있고, $\pi$가 바뀌지 않더라도 `policy-stable` 플래그는 계속 거짓이 되어 알고리즘이 종료되지 않는다.

가장 깔끔한 수정은 동률 처리에 일관성을 부여하는 것이다. 예를 들어 정책 개선에서 $\arg\max$가 옛 행동을 후보에 포함하면 옛 행동을 그대로 유지하도록 한다. 의사코드 3단계의 해당 부분을 다음과 같이 바꾸면 된다.

```
For each s in S:
    old-action ← π(s)
    candidate-set ← argmax_a Σ_{s',r} p(s',r | s,a)[r + γV(s')]
    if old-action ∈ candidate-set:
        π(s) ← old-action          # 동률이면 옛 행동 유지
    else:
        π(s) ← any element of candidate-set
        policy-stable ← false
```

이렇게 두면 정책이 더 이상 엄격하게 개선되지 않는 순간 `policy-stable`이 참으로 남아 알고리즘이 멈춘다. 동률 처리를 일관되게 하지 못하는 환경에서는, 새 행동의 추정 가치가 옛 행동의 추정 가치보다 임계값 $\theta$ 이상 클 때만 정책을 갱신하도록 두는 것도 같은 효과를 낸다.

## Exercise 4.5: 행동가치를 위한 정책 반복

$q_*$를 직접 계산하는 정책 반복은 본문 의사코드의 $V$를 $Q$로, 다음 상태의 평가에서 $V(s')$ 대신 $Q(s', \pi(s'))$를 사용하도록 바꾸면 된다. 정책 개선 단계도 단순한 $\arg\max_a Q(s, a)$ 형태가 된다.

```
1. Initialization
   Q(s, a) ∈ R 와 π(s) ∈ A(s) 를 모든 s ∈ S, a ∈ A(s) 에 대해 임의로 초기화
   Q(terminal, ·) = 0

2. Policy Evaluation
   Loop:
       Δ ← 0
       Loop for each s ∈ S:
           Loop for each a ∈ A(s):
               q ← Q(s, a)
               Q(s, a) ← Σ_{s',r} p(s', r | s, a) [ r + γ Q(s', π(s')) ]
               Δ ← max(Δ, |q − Q(s, a)|)
   until Δ < θ            # θ 는 작은 양수

3. Policy Improvement
   policy-stable ← true
   For each s ∈ S:
       old-action ← π(s)
       π(s) ← argmax_a Q(s, a)            # Exercise 4.4 의 동률 규칙 적용
       If old-action ≠ π(s) (그리고 동률이 아니면) policy-stable ← false
   If policy-stable: return Q ≈ q*, π ≈ π*; else go to 2
```

$q$ 형태의 핵심 이점은 정책 개선 단계에서 모델 $p(s', r \mid s, a)$를 다시 호출할 필요가 없다는 점이다. $V$를 쓸 때는 다음 상태의 가치를 한 단계 펼쳐 봐야 하지만, $Q$를 쓰면 그 자리에서 $\arg\max$만 취하면 된다. 이 차이는 모델이 없는 강화학습 알고리즘으로 자연스럽게 이어진다.

## Exercise 4.6: $\varepsilon$-soft 정책 제약

정책 $\pi$가 $\varepsilon$-soft라는 말은 모든 상태와 행동에 대해 $\pi(a \mid s) \geq \varepsilon / |\mathcal{A}(s)|$가 성립한다는 뜻이다. 이 제약을 정책 반복에 반영하면 다음과 같은 변화가 생긴다.

**3단계 정책 개선.** 더 이상 결정적 탐욕 정책을 만들 수 없다. 대신 각 상태에서 $\arg\max$ 행동에 큰 확률 $1 - \varepsilon + \varepsilon / |\mathcal{A}(s)|$를 부여하고, 나머지 행동에 각각 $\varepsilon / |\mathcal{A}(s)|$를 분배한다. 이렇게 만든 새 정책은 본문에서 본 $\varepsilon$-greedy 형태이며, 정책 개선 정리는 "$\varepsilon$-soft 정책들 가운데"에서 그대로 성립한다.

**2단계 정책 평가.** 정책이 더 이상 결정적이지 않으므로 갱신식에 정책 확률 $\pi(a \mid s)$가 명시적으로 들어와야 한다.

$$V(s) \leftarrow \sum_a \pi(a \mid s) \sum_{s', r} p(s', r \mid s, a)\bigl[r + \gamma V(s')\bigr].$$

행동가치 형태로 작업한다면 $Q(s, a) \leftarrow \sum_{s', r} p(s', r \mid s, a)[r + \gamma \sum_{a'} \pi(a' \mid s')\, Q(s', a')]$가 된다.

**1단계 초기화.** 초기 정책이 $\varepsilon$-soft 조건을 만족해야 한다. 가장 단순하게는 $\pi(a \mid s) = 1/|\mathcal{A}(s)|$인 균등 무작위 정책으로 시작하면 어떤 $\varepsilon \in [0, 1]$에 대해서도 조건을 만족한다.

이 변형의 결과는 일반적인 $\pi_*$가 아니라 $\varepsilon$-soft 정책 클래스 안의 최적 정책이다. 즉 모든 상태에서 적어도 $\varepsilon / |\mathcal{A}(s)|$의 탐험 확률을 유지한다는 제약 아래의 최적해를 찾는다.

## Exercise 4.7 (programming): 변형된 잭의 자동차 대여

본문 잭의 자동차 대여 문제에 두 가지 비선형 요소를 더한다.

1. 지점 1의 직원 한 명이 매일 밤 차 한 대를 지점 2로 무료로 옮긴다. 그 외의 이동은 한 대당 \$2이다.
2. 야간 이동을 마친 뒤 한 지점에 차가 10대를 넘으면 추가 주차장 사용료 \$4가 발생한다. 두 지점 모두 초과하면 \$8을 내야 한다.

이런 비선형성과 임의의 동역학은 동적 계획법이 잘 다루는 영역이다. 아래 구현은 [`figures.ipynb`](figures.ipynb)의 정책 반복 코드를 변형 비용 함수만 바꿔 그대로 활용한다.

````python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import BoundaryNorm
from scipy.stats import poisson

# 매개변수
MAX_CARS = 20
MAX_MOVE = 5
RENTAL_REWARD = 10
MOVE_COST = 2
PARKING_PENALTY = 4
PARKING_LIMIT = 10
GAMMA = 0.9
POISSON_UPPER = 11

LAMBDA_REQUEST = (3, 4)
LAMBDA_RETURN = (3, 2)


def poisson_pmf(lam: float, upper: int = POISSON_UPPER) -> np.ndarray:
    pmf = poisson.pmf(np.arange(upper + 1), lam)
    pmf[-1] += 1.0 - pmf.sum()
    return pmf


def location_dynamics(lam_req: float, lam_ret: float) -> tuple[np.ndarray, np.ndarray]:
    req_pmf = poisson_pmf(lam_req)
    ret_pmf = poisson_pmf(lam_ret)
    expected_reward = np.zeros(MAX_CARS + 1)
    transition = np.zeros((MAX_CARS + 1, MAX_CARS + 1))
    for start in range(MAX_CARS + 1):
        for req, p_req in enumerate(req_pmf):
            rented = min(start, req)
            expected_reward[start] += p_req * rented * RENTAL_REWARD
            after_rent = start - rented
            for ret, p_ret in enumerate(ret_pmf):
                end = min(after_rent + ret, MAX_CARS)
                transition[start, end] += p_req * p_ret
    return expected_reward, transition


REWARD_1, TRANS_1 = location_dynamics(LAMBDA_REQUEST[0], LAMBDA_RETURN[0])
REWARD_2, TRANS_2 = location_dynamics(LAMBDA_REQUEST[1], LAMBDA_RETURN[1])


def move_cost(action: int) -> int:
    """수정 1: 지점 1 -> 지점 2 방향의 첫 한 대는 무료."""
    if action > 0:
        return MOVE_COST * max(action - 1, 0)
    return MOVE_COST * abs(action)


def parking_cost(cars1: int, cars2: int) -> int:
    """수정 2: 야간에 한 지점에 10대 초과면 지점당 $4."""
    cost = 0
    if cars1 > PARKING_LIMIT:
        cost += PARKING_PENALTY
    if cars2 > PARKING_LIMIT:
        cost += PARKING_PENALTY
    return cost


def expected_update(state: tuple[int, int], action: int, V: np.ndarray) -> float:
    cars1 = min(state[0] - action, MAX_CARS)
    cars2 = min(state[1] + action, MAX_CARS)
    if cars1 < 0 or cars2 < 0:
        return -np.inf
    immediate = REWARD_1[cars1] + REWARD_2[cars2] - move_cost(action) - parking_cost(cars1, cars2)
    next_value = TRANS_1[cars1] @ V @ TRANS_2[cars2]
    return immediate + GAMMA * next_value


def policy_evaluation(policy: np.ndarray, V: np.ndarray, theta: float = 1e-3) -> np.ndarray:
    while True:
        delta = 0.0
        for i in range(MAX_CARS + 1):
            for j in range(MAX_CARS + 1):
                old = V[i, j]
                V[i, j] = expected_update((i, j), int(policy[i, j]), V)
                delta = max(delta, abs(old - V[i, j]))
        if delta < theta:
            return V


def policy_improvement(policy: np.ndarray, V: np.ndarray) -> tuple[np.ndarray, bool]:
    stable = True
    for i in range(MAX_CARS + 1):
        for j in range(MAX_CARS + 1):
            old_action = policy[i, j]
            actions = list(range(-min(MAX_MOVE, j), min(MAX_MOVE, i) + 1))
            q_values = [expected_update((i, j), a, V) for a in actions]
            best_idx = int(np.argmax(q_values))
            new_action = actions[best_idx]
            # Exercise 4.4 의 권고대로 동률이면 옛 행동을 유지
            if old_action in actions and abs(q_values[actions.index(int(old_action))] - q_values[best_idx]) < 1e-6:
                new_action = int(old_action)
            if old_action != new_action:
                stable = False
            policy[i, j] = new_action
    return policy, stable


V = np.zeros((MAX_CARS + 1, MAX_CARS + 1))
policy = np.zeros_like(V, dtype=int)
policies = [policy.copy()]
while True:
    V = policy_evaluation(policy, V)
    policy, stable = policy_improvement(policy, V)
    policies.append(policy.copy())
    if stable:
        break

# 시각화
fig = plt.figure(figsize=(14, 9))
fig.suptitle("Exercise 4.7: Modified Jack's car rental")

policy_levels = np.arange(-MAX_MOVE - 0.5, MAX_MOVE + 1.5)
policy_cmap = plt.get_cmap("RdBu_r", len(policy_levels) - 1)
policy_norm = BoundaryNorm(policy_levels, policy_cmap.N)

for idx, pi in enumerate(policies[: min(5, len(policies))]):
    ax = fig.add_subplot(2, 3, idx + 1)
    im = ax.imshow(pi, origin="lower", cmap=policy_cmap, norm=policy_norm)
    ax.set_title(rf"$\pi_{idx}$")
    ax.set_xlabel("#Cars at second location")
    ax.set_ylabel("#Cars at first location")
    fig.colorbar(im, ax=ax, ticks=range(-MAX_MOVE, MAX_MOVE + 1), shrink=0.8)

ax = fig.add_subplot(2, 3, 6, projection="3d")
X, Y = np.meshgrid(np.arange(MAX_CARS + 1), np.arange(MAX_CARS + 1))
ax.plot_surface(X, Y, V, cmap="viridis", edgecolor="none")
ax.set_title(r"$V_{\pi_*}$")
ax.set_xlabel("#Cars at second location")
ax.set_ylabel("#Cars at first location")

plt.tight_layout()
plt.show()
````

**해석.** 무료 셔틀 덕분에 지점 1에서 지점 2로 차 한 대를 옮기는 일은 "공짜 행동"이 되어, 같은 자본 위치에서도 본문 정책보다 양의 이동 영역이 살짝 넓어진다. 반대로 주차 비용은 두 지점 어느 쪽이든 차 수가 10대를 넘는 일을 강하게 억제하므로, 수렴한 정책은 양쪽 모두 10대 부근에 머물도록 차를 더 부지런히 분산시킨다. 본문 결과를 먼저 재현하려면 `move_cost`와 `parking_cost`를 원래 형태로 되돌려 비교하면 된다.

## Exercise 4.8: 도박사 정책의 계단 모양

$p_h = 0.4$이면 매 베팅의 기대값은 음수이므로, 합리적 도박사는 가능한 한 적은 횟수의 베팅으로 결판을 내려고 한다. 자본 50에서 50을 모두 걸면 한 번에 100에 도달할 확률이 $p_h = 0.4$이다. 어떤 단계적 "안전한" 전략도 같은 확률을 따라잡지 못한다. 자본이 늘어날수록 작은 베팅이 누적되며 결국 0에 도달할 확률이 더 커지기 때문이다. 그래서 자본 50에서는 전부를 거는 것이 최적이다.

자본 51에서는 사정이 다르다. 가능한 베팅의 상한이 $\min(51, 49) = 49$로 줄어들어, 한 번의 승부로 100에 닿는 베팅(49)을 했을 때 기대 수익은 $0.4 \cdot V(100) + 0.6 \cdot V(2) = 0.4 + 0.6\,V(2)$이다. $V(2)$는 매우 작으므로 이 값은 $V(50) = 0.4$를 넘지 못한다. 반면 1만 걸면 $0.4\,V(52) + 0.6\,V(50)$이 되어, $V(52)$가 $V(50)$보다 약간만 커도 더 큰 기대 수익이 된다. 결국 51에서는 "이왕이면 50으로 한 발 물러서서 그곳에서 모두 걸자"는 전략이 더 우월하다.

이 같은 논리가 25, 75 같은 자본에서도 비슷하게 작동한다. 25에서 25를 걸면 50으로 가고, 거기서 다시 한 번에 100에 도달할 수 있다. 75에서 25를 걸면 100에 직행한다. 따라서 정책에 "큰 베팅이 필요한 자본"의 사다리가 만들어지고, 그 사이의 자본에서는 사다리의 다음 칸을 향해 작은 베팅을 한다. 결과적으로 정책은 1, 25, 50, 75 같은 위치를 봉우리로 하는 자기 닮음(self-similar) 계단 모양을 띤다.

## Exercise 4.9 (programming): $p_h = 0.25$, $0.55$에 대한 가치 반복

[`figures.ipynb`](figures.ipynb)에서 정의한 가치 반복을 두 확률값에 대해 다시 돌려, 가치 함수의 수렴과 최종 정책의 모양을 비교한다.

````python
import numpy as np
import matplotlib.pyplot as plt

GOAL = 100
THETA = 1e-9


def value_iteration(p_h: float, goal: int = GOAL, theta: float = THETA) -> tuple[np.ndarray, np.ndarray, list[np.ndarray]]:
    V = np.zeros(goal + 1)
    V[goal] = 1.0
    snapshots: dict[int, np.ndarray] = {}
    sweep = 0
    while True:
        sweep += 1
        delta = 0.0
        for s in range(1, goal):
            actions = np.arange(0, min(s, goal - s) + 1)
            returns = p_h * V[s + actions] + (1 - p_h) * V[s - actions]
            best = returns.max()
            delta = max(delta, abs(V[s] - best))
            V[s] = best
        if sweep in (1, 2, 3, 32):
            snapshots[sweep] = V.copy()
        if delta < theta:
            break
    policy = np.zeros(goal + 1, dtype=int)
    for s in range(1, goal):
        actions = np.arange(0, min(s, goal - s) + 1)
        returns = p_h * V[s + actions] + (1 - p_h) * V[s - actions]
        # 동률을 흡수해 가장 작은 베팅을 고른다.
        policy[s] = actions[np.argmax(np.round(returns, 5))]
    ordered = [snapshots[k] for k in sorted(snapshots)]
    return V, policy, ordered


fig, axes = plt.subplots(2, 2, figsize=(12, 9))
fig.suptitle("Exercise 4.9: Gambler's problem for two coin biases")

for col, p_h in enumerate([0.25, 0.55]):
    V, policy, snapshots = value_iteration(p_h)
    labels = ["sweep 1", "sweep 2", "sweep 3", "sweep 32"]
    for snap, label in zip(snapshots, labels):
        axes[0, col].plot(np.arange(1, GOAL), snap[1:GOAL], label=label)
    axes[0, col].plot(np.arange(1, GOAL), V[1:GOAL], color="black", label="final")
    axes[0, col].set_title(rf"$p_h = {p_h}$, value estimates")
    axes[0, col].set_xlabel("Capital")
    axes[0, col].set_ylabel("Value")
    axes[0, col].legend()

    axes[1, col].step(np.arange(1, GOAL), policy[1:GOAL], where="mid")
    axes[1, col].set_title(rf"$p_h = {p_h}$, final policy (stake)")
    axes[1, col].set_xlabel("Capital")
    axes[1, col].set_ylabel("Stake")

plt.tight_layout()
plt.show()
````

**해석.**

- $p_h = 0.25$: 도박사가 더 큰 불리함을 안고 있다. 가치 함수는 $p_h = 0.4$의 경우와 비슷한 자기 닮음 계단 모양을 보이지만 전체적으로 값이 훨씬 낮고, 같은 "한 번에 결판" 전략이 더 도드라진다. 자본 50, 25, 75 같은 사다리 위치에서 모두 걸기 행동이 더 강하게 나타난다.
- $p_h = 0.55$: 매 베팅의 기대값이 양수이므로 합리적 도박사는 베팅 횟수를 늘려 큰 수의 법칙에 기댄다. 가치 함수가 자본의 매끄러운 증가 함수에 가까워지고, 최적 정책은 거의 모든 자본에서 "가능한 가장 작은 베팅"인 1을 고르는 형태가 된다. 동률이 많아 다른 최적 정책도 존재하지만, 위 코드는 동률 가운데 가장 작은 베팅을 선택하므로 1로 평탄한 그래프가 그려진다.

$\theta \to 0$의 안정성에 대해서는, 두 경우 모두 충분히 작은 $\theta$에서 가치 함수가 한 점으로 수렴하므로 결과는 안정적이다. 다만 정책 그래프는 동률 처리 방식에 따라 다르게 보일 수 있다는 점이 미묘한 차이로 남는다.

## Exercise 4.10: 행동가치에 대한 가치 반복

본문 식 (4.10)은 $v_*$에 대한 벨만 최적 방정식을 갱신 규칙으로 옮긴 것이다. $q_*$에 대한 벨만 최적 방정식

$$q_*(s, a) = \sum_{s', r} p(s', r \mid s, a)\Bigl[r + \gamma \max_{a'} q_*(s', a')\Bigr]$$

을 같은 방식으로 옮기면 행동가치 가치 반복 갱신이 나온다.

$$q_{k+1}(s, a) \doteq \sum_{s', r} p(s', r \mid s, a)\Bigl[r + \gamma \max_{a'} q_k(s', a')\Bigr]. \tag{4.10$'$}$$

임의의 $q_0$에서 출발해 이 갱신을 반복하면 $v_*$가 존재하는 같은 조건 아래에서 $q_k \to q_*$가 된다. 수렴 후에는 모델 없이 단순히 $\pi_*(s) = \arg\max_a q_*(s, a)$로 최적 정책을 뽑을 수 있다는 점이 $v_*$ 형태와의 큰 차이이다.
