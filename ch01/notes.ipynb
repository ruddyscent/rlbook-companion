# 1 소개 (Introduction)

## 1.1 강화학습이란 무엇인가

**강화학습(reinforcement learning)**은 수치적 보상 신호를 최대화하기 위해, 어떤 상황에서 어떤 행동을 취해야 할지를 학습하는 문제이다. 이를 수행하는 주체를 **에이전트(agent)**라 부른다. 에이전트에게 "어떤 행동이 정답인가"는 알려 주지 않으며, 에이전트는 직접 행동을 시도해 보고 그 결과로부터 어떤 행동이 가장 큰 보상을 가져오는지 스스로 발견해야 한다. 그리고 한 행동의 영향은 즉각적인 보상에 그치지 않고, 그 이후의 모든 상황과 보상에까지 미친다.

이 두 가지 특징, 즉 **시행착오 탐색(trial-and-error search)**과 **지연된 보상(delayed reward)**은 강화학습을 지도학습이나 비지도학습과 구분 짓는 가장 본질적인 성질이다. 지도학습은 외부 감독자가 정답 라벨을 제공하는 학습이고, 비지도학습은 라벨 없는 데이터에서 숨겨진 구조를 찾는 학습이다. 강화학습은 이 둘 어느 쪽에도 속하지 않으며, 기계학습의 **세 번째 패러다임**으로 자리한다.

강화학습이라는 단어는 세 가지 뜻을 동시에 가진다: **해결해야 할 문제** 자체, 그 문제에 잘 작동하는 **해법 방법론의 집합**, 그리고 이 문제와 해법을 연구하는 **학문 분야**. 이 셋을 하나의 이름으로 부르면 편리하지만, 문제와 해법은 개념적으로 구분해야 한다.

강화학습은 본질적으로 **탐험(exploration)과 활용(exploitation) 사이의 균형 문제**를 안고 있다. 에이전트는 이미 좋다고 알려진 행동을 활용해 보상을 얻는 동시에, 더 좋은 행동을 찾으려고 새로운 행동을 탐험해야 한다. 어느 하나만으로는 과제를 풀 수 없으며, 둘 사이에서 균형을 잡는 일이 강화학습 고유의 도전이다.

강화학습의 또 다른 핵심 특징은 **목표 지향적 에이전트가 불확실한 환경과 상호작용하는 과정**을 잘라내거나 단순화하지 않고 그대로 하나의 문제로 다룬다는 점이다. 많은 기계학습 접근법이 더 큰 그림 속에서의 역할을 고려하지 않은 채 부분 문제만을 다루는 데 반해, 강화학습은 에이전트가 명확한 목표를 향해 환경을 감지하고 행동으로 환경에 영향을 미치는 **그 상호작용 과정 전체**를 하나의 문제로 삼는다.

강화학습은 공학, 통계학, 최적화뿐 아니라 **심리학과 신경과학**과도 깊이 맞닿아 있다. 모든 기계학습 방식 가운데 강화학습이 인간과 동물의 학습 방식과 가장 닮았으며, 핵심 알고리즘 다수도 생물학적 학습 시스템에서 아이디어를 끌어왔다.

## 1.2 사례

강화학습이 적용되는 사례는 분야를 가리지 않는다.

- 체스 마스터의 수 선택 — 직관적 판단과 상대의 응수에 대한 앞선 추론을 결합해 최선의 다음 수를 고른다.
- 정유 공장의 적응 제어기 — 서로 상충하는 수율, 비용, 품질 사이에서 최선의 균형점을 실시간으로 찾아낸다.
- 갓 태어난 가젤의 보행 학습 — 태어난 직후 시행착오를 거치며 짧은 시간 안에 효율적인 움직임을 익힌다.
- 청소 로봇의 의사결정 — 배터리 잔량과 지금까지 쌓인 경험을 바탕으로, 청소를 계속할지 충전하러 돌아갈지를 스스로 판단한다.
- 아침 식사 준비 — 재료 선택부터 조리와 상차림까지, 수많은 세부 목표와 행동이 순서를 맞춰 이어지는 일상적 과제다.

이들의 공통점은, 능동적이고 의사결정을 내리는 에이전트가 **불확실한 환경**과 상호작용하면서 **명시적 목표**를 향해 학습한다는 점이다.

## 1.3 강화학습의 네 가지 구성 요소

에이전트와 환경 외에, 강화학습 시스템은 다음 네 가지 하위 요소를 포함한다.

**정책(policy).** 각 상태에서 에이전트가 어떤 행동을 선택할지를 결정하는 규칙으로, 상태마다 행동 하나를 지정하거나 행동별 선택 확률을 부여한다. 심리학의 자극–반응 규칙과 같은 개념이다. 정책은 강화학습 에이전트의 핵심이며, 정책만으로 에이전트의 행동이 완전히 결정된다.

**보상 신호(reward signal).** 강화학습 문제의 **목표를 정의**한다. 매 시점 환경은 단일한 수치를 에이전트에게 전달하며, 에이전트의 유일한 목적은 장기적으로 받을 보상의 합을 최대화하는 것이다. 보상은 어떤 행동이 즉각적으로 좋은지를 알려 주는 신호이며, 무엇이 옳은 행동인지를 판단하는 일차적 기준이 된다.

**가치 함수(value function).** 보상 신호가 즉각적인 가치를 나타낸다면, 가치 함수는 **장기적인 가치**를 나타낸다. 어떤 상태의 가치는 그 상태에서 출발했을 때 앞으로 받을 수 있는 보상의 기댓값이다. 보상은 환경이 직접 알려 주지만, 가치는 에이전트가 경험을 쌓으며 스스로 추정해야 한다는 점에서 둘은 성격이 다르다. 강화학습에서 행동을 결정할 때는 즉각적 보상이 아니라 가치를 기준으로 삼아야 한다.

**환경 모델(model).** 일부 강화학습 시스템은 환경의 작동 방식을 본뜬 모델을 사용한다. 모델은 주어진 상태와 행동에 대해 다음 상태와 보상을 예측하며, 에이전트는 이를 활용해 실제로 경험하기 전에 행동의 결과를 미리 가늠하는 **계획(planning)**을 수행할 수 있다. 모델을 사용하는 방법을 **모델 기반(model-based)** 강화학습, 모델 없이 시행착오에만 의존하는 방법을 **모델을 사용하지 않는(model-free)** 강화학습이라 부른다.

## 1.4 한계와 적용 범위

강화학습은 **상태(state)**라는 개념에 크게 의존한다. 상태는 정책과 가치 함수가 작동하는 기반이며, 모델에서는 입력으로도 출력으로도 쓰인다. 다만 상태 신호를 어떻게 구성할 것인지는 별도의 어려운 문제이고, 강화학습이 집중하는 질문은 **주어진 상태에서 어떻게 의사결정을 학습할 것인가**이다.

강화학습은 가치 함수를 활용한 학습 방법을 중심에 두며, 유전 알고리즘처럼 정책을 직접 탐색하는 **진화적 방법**과는 선을 긋는다. 진화적 방법은 학습 중에 환경과의 상호작용에서 얻은 정보를 학습에 반영하지 않는다는 점에서 강화학습과 근본적으로 다르다.

## 1.5 확장 예제: 틱택토

틱택토는 강화학습이 고전적 방법들과 어떻게 다른지를 잘 보여 주는 예제다. 미니맥스(minimax)는 상대가 항상 최선의 수를 둔다고 가정해 행동을 선택하므로, 실수를 하는 상대의 약점을 파고들지 못한다. 동적 계획법(dynamic programming)은 상대 행동의 완전한 확률 모델을 요구하는데, 이런 모델은 대개 미리 주어지지 않는다.

강화학습은 **각 보드 상태에서의 승리 확률을 가치 함수로 추정**한다. 가치 테이블은 다음과 같이 초기화한다. X가 세 개 나란히 놓인 승리 상태는 1, 패배·무승부 상태는 0, 나머지는 0.5로 설정한다.

학습 도중 에이전트는 대부분 가장 가치가 높은 수를 두는 **탐욕적 선택**을 하지만, 이따금 무작위로 수를 두는 **탐험적 선택**도 한다. 탐험적 수에서는 학습이 일어나지 않으며, 탐욕적 수를 둘 때마다 다음 상태의 가치로 직전 상태의 가치를 갱신한다.

$$V(S_t) \leftarrow V(S_t) + \alpha \big[ V(S_{t+1}) - V(S_t) \big]$$

$S_t$는 현재 상태, $S_{t+1}$는 다음 상태, $\alpha$는 두 상태의 가치 차이를 현재 추정값에 얼마나 반영할지를 조절하는 단계 크기 매개변수이다. 이 갱신 방식은 **시간 차분 학습(temporal-difference learning)**의 한 형태로, 책 전체에서 중심적인 역할을 한다.

$\alpha$를 점차 줄여 가면 이 방법은 고정된 상대에 대해 각 상태에서의 참 승리 확률, 즉 최적 정책으로 수렴한다. $\alpha$를 완전히 0으로 만들지 않으면 상대의 전략이 서서히 바뀌어도 잘 따라갈 수 있다.

진화적 방법과의 차이도 눈여겨볼 만하다. 진화적 방법은 정책 전체를 고정한 채 게임 결과만으로 다음 정책을 고르므로, 게임 중 어떤 수가 결과를 좌우했는지를 전혀 활용하지 못한다. 반면 가치 함수 기반 방법은 각 상태를 개별적으로 평가하고 플레이 중 얻은 정보를 바로 학습에 반영한다.

틱택토는 단순한 예에 불과하지만 강화학습의 적용 범위는 훨씬 넓다. 연속적인 과제, 크기가 다양한 보상, 상태 공간이 매우 크거나 무한한 문제에도 똑같이 적용된다. 테사우로(Tesauro)는 이 접근법을 인공 신경망과 결합해 약 $10^{20}$개의 상태를 가진 백게이먼을 학습시켜 세계 최고 수준의 인간 플레이어를 넘어섰다. 틱택토 플레이어가 상대의 모델 없이 실제 플레이 양상에 직접 적응하듯, 이는 **모델을 사용하지 않는** 방식의 전형적인 예이기도 하다.

### Exercise 1.1: 자신을 상대로 한 대국

강화학습 알고리즘이 틱택토에서 자기 자신을 상대로 학습하는 상황을 생각해 본다. 대칭성에 관한 논의는 다음 문제([Exercise 1.2](#exercise-1-2))에서 다루기로 하고, 여기서는 자신을 상대로 한 대국이 학습 결과에 미치는 영향에만 초점을 맞춘다.

**탐험과 활용.** 학습 과정 내내 알고리즘은 다양한 행동을 시도하며 탐험을 이어 간다. 자신을 상대로 한 대국에서는 상대편 또한 동일하게 탐험을 수행하는 또 하나의 학습자이므로, 양쪽 모두 자신의 경험으로부터 점진적으로 배운다.

**최적 전략으로의 수렴.** 틱택토는 결정론적이고 상태 공간이 유한하므로, 학습이 충분히 진행되면 알고리즘은 가능한 모든 상태를 두루 거친다. 그리고 그 과정에서 게임의 최적 정책으로 수렴해 간다. 틱택토의 최적 전략은 잘 알려져 있고, 선공이든 후공이든 "이기거나 적어도 비기는" 형태로 요약할 수 있다.

자신을 상대로 대국하는 환경에서도 강화학습 알고리즘은 결국 최적 정책을 학습할 가능성이 높다. 학습 결과는 승리 가능성을 최대로 끌어올리거나 적어도 무승부를 강제하는 정책으로 모인다.

### Exercise 1.2: 대칭성

틱택토 보드에는 회전과 반사에 의한 대칭이 존재한다. 이러한 구조를 학습 과정에 어떻게 반영할 수 있는지, 그리고 그것이 어떤 이점을 가져오는지를 살펴본다.

#### 1. 대칭성을 활용하는 방법

- **상태 표현:** 학습 알고리즘은 대칭적으로 동등한 위치들을 별개의 상태로 취급하지 않고 동일한 상태로 묶을 수 있다. 예컨대 어떤 보드 배치가 다른 배치의 회전 또는 반사 결과라면, 두 배치를 같은 상태로 본다.
- **가치 공유:** 가치 추정값을 갱신할 때, 대칭으로 동등한 상태끼리 그 값을 공유한다. 한 상태를 유리하다고 보았다면 그와 대칭인 상태에도 같은 평가를 그대로 적용할 수 있다.

#### 2. 학습 과정의 개선 효과

- **효율성:** 알고리즘이 따로 탐색하고 학습해야 하는 상태 수가 줄어들기 때문에 전체 계산 비용이 줄어든다.
- **일반화:** 대칭으로 동등한 위치들에서 배운다는 것은 곧 그 동등한 상태 전반에 지식이 퍼져 나간다는 뜻이다. 그 결과 알고리즘이 게임을 한층 견고하고 일반적인 방식으로 이해한다.

#### 3. 상대가 대칭성을 활용하지 않을 때

상대가 대칭성을 이용하지 않더라도, 학습자가 이를 활용하는 편은 여전히 유리하다. 효율성과 일반화 측면의 이득이 그대로 유지되기 때문이다. 다만 그 효과의 크기는 상대 전략의 특성과 게임 안에 대칭 구조가 얼마나 자주 등장하는지에 따라 달라진다.

#### 4. 대칭으로 동등한 위치는 같은 가치를 가져야 하는가

반드시 그렇지는 않다. 게임의 성질상 대칭 위치는 비슷한 가치를 갖는 경우가 많지만, 맥락에 따라 최적 수가 달라지는 상황도 있다. 따라서 알고리즘은 이러한 미묘한 차이까지 구분하여 각 상태의 구체적 맥락에 맞는 가치를 부여하도록 학습해야 한다.

### Exercise 1.3: 탐욕적 플레이

항상 자신이 가장 좋다고 평가한 수만 두는 탐욕적 플레이어의 학습 성능이 그렇지 않은 플레이어보다 나은지 따져 본다.

**더 잘 학습할 가능성.** 탐욕적 전략은 지금까지 모은 지식으로 보아 즉각적인 가치가 가장 높은 수를 일관되게 고른다. 학습 알고리즘이 이미 정확한 가치 함수에 충분히 수렴해 있다면, 탐욕적 플레이어는 매 상황에서 최적의 수를 두며 좋은 성과를 낼 수 있다.

**학습을 저해할 위험.** 그러나 순수한 탐욕적 전략은 학습 과정에서 충분한 탐험을 수행하지 않을 위험이 크다. 새롭고 더 나은 전략을 발견하려면 탐험이 필수적인데, 다양한 수를 시도해 보지 않으면 장기적으로 우수한 전략을 놓치게 되고 결국 차선책에 머무는 플레이를 학습하게 된다.

탐욕적 플레이에는 세 가지 문제가 있다.

- **탐험 부족:** 알고리즘이 더 좋은 대안을 발견할 기회를 잃는다. 학습이 아직 정확한 가치 함수에 수렴하지 못한 단계라면 이 문제는 더욱 심각해진다.
- **국소 최적해:** 현재 지식 기준으로 가장 좋아 보이는 수에 안주한 결과, 전역적으로 더 나은 전략을 놓치고 국소 최적해에 갇힐 수 있다. 수들 사이의 상호작용이 복잡할수록 학습은 더욱 정체된다.
- **적응력 저하:** 상대 전략의 변화나 예기치 못한 상황에 적응하기 어렵다. 더 유연한 정책을 학습하려면 당장의 최선책에서 벗어나는 시도, 즉 탐험이 필요하다.

탐욕적 플레이어는 이미 최적 정책에 수렴해 있다면 좋은 성과를 내지만, 그렇지 못한 경우에는 차선의 전략에 갇힐 가능성이 크다. 효과적으로 학습하려면 활용과 탐험 사이에서 균형을 잡아야 한다.

### Exercise 1.4: 탐험 결과로부터의 학습

원래 알고리즘은 탐욕적 수를 둘 때만 가치를 갱신하고, 탐험적 수를 둘 때는 갱신하지 않는다. 탐험적 수에서도 가치를 갱신하면 어떻게 달라질까.

**탐험적 수에서도 갱신하는 경우.** 가치 추정값은 탐욕적 선택과 탐험적 선택이 뒤섞인 실제 플레이, 즉 ε-탐욕 정책 아래에서의 승리 확률로 수렴한다.

**탐험적 수에서는 갱신하지 않는 경우.** 가치 추정값은 탐험 없이 순수하게 탐욕적으로 플레이했을 때의 승리 확률로 수렴한다.

**어느 쪽이 나은가.** 실제 플레이에서 탐험을 계속한다면, 탐험적 수까지 반영한 쪽이 실제 행동과 일치하므로 더 많은 승리로 이어진다. 탐험을 배제하고 학습한 가치는 매번 최선의 수만 두었을 때의 승률을 정확히 나타내지만, 탐험을 병행하는 실제 상황과는 맞지 않는다.

### Exercise 1.5: 그 밖의 개선 방안

강화학습 플레이어 자체를 개선하는 방법과, 틱택토 문제를 더 효과적으로 해결하기 위한 방법으로 나누어 살펴본다.

#### 1. 강화학습 플레이어를 개선하는 방법

- **함수 근사:** 큰 상태 공간에서는 표 기반 방법이 비현실적이므로, 선형 함수 근사나 신경망 같은 함수 근사 기법을 도입한다. 이를 통해 한 번도 본 적 없는 상태에 대해서도 학습 결과를 일반화할 수 있다.
- **경험 재현:** 과거 경험을 저장해 두었다가 학습 시 무작위로 샘플링해 활용하는 기법으로, 학습을 더 안정적이고 효율적으로 만든다. 함수 근사 기법과 함께 자주 활용한다.
- **인간 전문가 지식의 활용:** 인간 전문가의 지식이나 기존 전략을 학습 초기에 반영하면, 강화학습이 사전 지식에서 출발해 정책을 점진적으로 다듬어 갈 수 있다.
- **몬테카를로 트리 탐색(MCTS):** 트리 기반 탐색 알고리즘으로 탐험과 활용을 자연스럽게 결합한다. 다양한 보드 게임에서 큰 성과를 거둔 기법이며, 틱택토 플레이어에도 응용할 수 있다.
- **동적 학습률:** 환경 변화에 따라 학습률을 적응적으로 조정하면 학습 과정을 개선할 수 있다. Adagrad나 Adam 같은 최적화 기법을 검토할 만하다.

#### 2. 틱택토 문제 자체를 더 잘 푸는 방법

- **최적 정책의 사전 계산:** 틱택토는 상태 공간이 작기 때문에, 모든 상태에 대해 최적 정책을 미리 계산해 둘 수 있다. 이렇게 하면 게임 중에 학습이 필요하지 않지만, 환경에 대한 완전한 지식을 전제로 한다는 한계가 있다.
- **게임 고유의 휴리스틱(heuristic):** 틱택토의 알려진 최적 전략을 바탕으로 도메인 특화 휴리스틱을 도입하면, 학습이 강력한 초기 정책에서 출발할 수 있다.
- **패턴 인식:** 승리나 패배와 결부되는 보드 패턴을 인식하는 기법을 활용하면, 알고리즘은 그 정보를 바탕으로 더 합리적인 의사결정을 내릴 수 있다.
- **앙상블 기법:** 여러 강화학습 모델이나 전략을 앙상블 방식으로 결합하면, 더 견고하고 믿을 만한 틱택토 플레이어를 만들 수 있다.

개선 전략의 선택은 문제의 특성과, 계산 비용 대비 성능 사이의 균형 등 구체적인 요구사항에 따라 달라진다.

## 1.6 요약

강화학습은 목표를 향해 행동하는 에이전트가 환경과의 상호작용을 통해 스스로 배우는 학습 방식이다. 에이전트에게 정답을 알려 주지 않는다는 점, 그리고 한 행동의 결과가 먼 미래의 보상에까지 영향을 미친다는 점이 다른 학습 방식들과의 근본적인 차이다. 에이전트–환경 상호작용은 **마르코프 결정 과정(Markov decision process, MDP)**으로 수학적으로 기술하며, 정책·보상·가치 함수·환경 모델이라는 네 요소를 중심으로 알고리즘이 구성된다.

## 1.7 초기 역사

강화학습의 역사는 본래 **두 개의 독립적인 흐름**이 한데 합쳐진 결과이며, 나중에 세 번째 흐름이 합류해 현대 강화학습을 완성했다.

### 최적 제어 흐름

**최적 제어(optimal control)**라는 용어는 1950년대 말에 등장했으며, 리처드 벨만(Richard Bellman)은 해밀턴-야코비 이론을 확장해 **동적 계획법**을 개발했다(Bellman, 1957). 벨만 방정식이라 불리는 함수 방정식은 최적 제어 문제를 체계적으로 풀 수 있는 수학적 기반이 되었다. 벨만은 또한 이산 확률 최적 제어 문제인 **마르코프 결정 과정**을 도입했고(Bellman, 1957), 로널드 하워드(Ronald Howard, 1960)는 MDP를 위한 **정책 반복(policy iteration)** 방법을 고안했다. 현대 강화학습의 이론과 알고리즘 대부분이 이에 뿌리를 둔다.

동적 계획법은 차원의 저주(curse of dimensionality) — 상태 변수 수에 따라 계산량이 지수적으로 증가하는 문제 — 로 어려움을 겪었지만, 그럼에도 확률 최적 제어 문제를 일반적으로 풀 수 있는 사실상 유일한 수단으로 오랫동안 자리했다.

### 시행착오 학습 흐름

시행착오 학습의 뿌리는 동물 학습 심리학이다. **에드워드 손다이크(Edward Thorndike)**의 **효과의 법칙(Law of Effect)**은 만족스러운 결과를 낳은 행동이 다시 선택될 가능성이 높아진다는 원리를 처음으로 명확히 했다. **앨런 튜링(Alan Turing)**은 1948년 보고서에서 효과의 법칙에 따라 작동하는 "쾌락-고통 시스템"을 설계했다.

초기의 학습 기계들도 주목할 만하다. **도널드 미치(Donald Michie)**는 1961년과 1963년에 틱택토를 학습하는 시스템 **MENACE**(Matchbox Educable Naughts and Crosses Engine)를 고안했다. 각 게임 상태에 해당하는 성냥갑에 행동별로 색깔이 다른 구슬을 담아두고, 게임 결과에 따라 구슬을 추가하거나 제거해 학습했다. 미치와 체임버스(Michie and Chambers, 1968)는 이후 막대 균형 잡기 문제에 강화학습 제어기 **BOXES**를 적용했는데, 이는 불완전한 지식 조건에서 강화학습이 성공한 대표적 초기 사례이다.

**학습 오토마톤(learning automata)** 연구도 이 흐름에 중요한 기여를 했다. 이는 $k$-팔 밴딧(k-armed bandit) 문제를 풀기 위한 방법으로, 러시아 수학자 체틀린(Tsetlin) 등의 연구에서 출발해 공학 분야에서 발전했다.

### 시간 차분 학습 흐름

시간 차분 학습의 기원은 동물 학습 심리학의 2차 강화인자(secondary reinforcer) 개념에 있다. **아서 새뮤얼(Arthur Samuel)**은 1959년 체커 프로그램에서 시간 차분 아이디어를 처음으로 구현했다. **이언 위튼(Ian Witten)**은 1977년 MDP를 푸는 적응형 제어기의 일부로 시간 차분 학습 규칙(현재의 표형 TD(0))을 최초로 발표했다.

서튼(Sutton)과 바토(Barto)는 1981년 시간 차분 학습과 시행착오 학습을 결합한 **액터-크리틱(actor-critic) 아키텍처**를 개발해 BOXES의 막대 균형 잡기 문제에 적용했다. 서튼(1988)은 시간 차분 학습을 제어와 분리해 독립적인 예측 방법으로 발전시키고 TD($\lambda$) 알고리즘을 도입했다.

### 세 흐름의 합류

1989년 **크리스 왓킨스(Chris Watkins)**가 Q-학습을 개발하면서 시간 차분 흐름과 최적 제어 흐름이 완전히 통합되었다. 1992년 테사우로의 백게이먼 프로그램 **TD-Gammon**의 눈부신 성공이 강화학습 분야에 큰 주목을 끌었다. 이후 시간 차분 알고리즘이 뇌의 도파민 뉴런 활동과 놀랍도록 유사하다는 사실이 밝혀지면서 신경과학 분야와의 접점도 넓어졌다.
